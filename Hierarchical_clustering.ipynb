```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn import datasets
```

```python
iris= datasets.load_iris()
```

```python
iris_data= pd.DataFrame(iris.data)
```

```python
iris_data.columns=iris.feature_names
```

```python
iris_data
```

```python
#standardization
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
```

```python
X_scaled=scaler.fit_transform(iris_data)
```

```python
X_scaled
```

```python
X_scaled.shape
```

```python
#Applying the PCA
from sklearn.decomposition import PCA
pca=PCA(n_components=2)#4 dimension 2 dimension me convert hoga
```

```python
pca
```

```python
pca_scaled=pca.fit_transform(X_scaled)
```

```python
pca_scaled
```

```python
plt.scatter(pca_scaled[:,0],pca_scaled[:,1])
```

```python
#Agglomerative Clustering
#To construct a dendogram
import scipy.cluster.hierarchy as sc#dendogram construct krne k liye hame scipy library use krna pdta hai
#plot the dendogram
plt.figure(figsize=(20,7))
plt.title("Dendogram")
#create dendogram
sc.dendrogram(sc.linkage(pca_scaled,method='ward'))
plt.xlabel("Sample Index")
plt.ylabel("Euclidean Distance")
```

```python
from sklearn.cluster import AgglomerativeClustering
cluster=AgglomerativeClustering(n_clusters=2, metric='euclidean', linkage='ward')
cluster.fit(pca_scaled)#affinity parameter is changed to metric
```

```python
cluster.labels_
```

```python
plt.scatter(pca_scaled[:,0],pca_scaled[:,1],c=cluster.labels_)
```
